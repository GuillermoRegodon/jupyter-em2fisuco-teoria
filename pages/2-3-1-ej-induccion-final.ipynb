## Vuelta a la inducción de Faraday en relatividad

* Volvemos al problema de una espira en movimiento relativo con un campo magnético generado, por ejemplo, por unos imanes permanentes:

```{figure} ../images/relat_10_1.png
:alt: relat_10_1.png
:width: 600px
SRI en Faraday.
```

En el SRI1, la espira está en movimiento con respecto al campo magnético que no varía con respecto al tiempo en cada punto del espacio. En el SRI2, la espira está quieta y el iman se desplaza, por lo que el campo magnético está variando en el tiempo en cada punto del espacio. La relación entre los dos sistemas de referencia es la de una velocidad relativa $\vec{v}$.


```{figure} ../images/relat_10_2.png
:alt: relat_10_2.png
:width: 300px
SRI1 y SRI2.
```

Para concretar las direcciones, vamos a suponer una región de campo magnético constante tanto en el tiempo como en el espacio en el SRI1. El campo electrico es nulo. En este caso la $fem$ se obtiene a través de la ley de inducción.

$$
fem = - \frac{d \Phi}{dt} = - B L v
$$

Como no hay campo eléctrico, y las cargas del conductor está en movimiento a velocidad $\vec{v}$, la fuerza que desplaza las cargas y genera la $fem$ debe ser la fuerza magnética.

Ahora, vamos a calcular cómo se transforman los campos electromagnéticos con el cambio de sistema de referencia.

$$
\vec{E} &= 0 \\
\vec{B} &= B \vec{u_x} \\
\vec{v} &= v \vec{u_y}
$$

En el sistema de refrencia SRI2, los campos transformados son:

$$
E'_{||} &= E_{||}\\
B'_{||} &= B_{||}\\
\vec{E'_\perp} &= \gamma(\vec{E_\perp} + \vec{v}\times\vec{B}) \\
\vec{B'_\perp} &= \gamma(\vec{B_\perp} - \frac{\vec{v}}{c^2}\times\vec{E}) \\
$$

Los campos en la dirección paralela a la velocidad $\vec{v}$ es la dirección $y$ no cambian, es decir, siguen siendo 0.

$$
E'_{||} &= E_y = 0\\
B'_{||} &= B_y = 0\\
$$

La dirección perpendicular incluye la dirección $x$. Empezamos con el campo magnético:

$$
\vec{B'_\perp} = \gamma B \vec{u_x}
$$

Tiene un valor determinado (*vale 0 en la región externa a esta región*), pero como la espira está quieta en SRI2, no genera fuerza en las cargas del conductor. Genera una variación de flujo magnético. Si la región tiene una anchura $L$:

$$
fem' = - \frac{d \Phi'}{dt} = - B' L v = - \gamma B L v
$$

Como decimos, esta $fem$ no esta generada por una fuerza magnética. Veamos el campo eléctrico:

$$
\vec{E'_\perp} = \gamma \left( (v \vec{u_y})\times(B \vec{u_x}) \right) = - \gamma B v \vec{u_z}
$$

Dentro de la región de campo magnético (*vale 0 en la región externa a dicha región*). Observemos que si integramos el campo eléctrico (*no conservativo*) a lo largo de la linea del conductor afectado por el campo eléctrico, de longitud $L$, obtenemos justo la expresión anterior.

$$
\oint \vec{E'}\cdot\vec{dl} = \int_L \vec{E'}\cdot\vec{dl} = - \gamma B L v
$$

Dando la misma expresión de la fem a partir de la variación del flujo de campo magnético. Caundo la velocidad no es relativista, $$\gamma {\approx} 1$$ y obtenemos el resultado conocido del electromagnetismo.

```{note}
Observamos que, **en el sistema de referencia SRI1 la fuerza que genera la $fem$ es magnética, mientras que en el sistema de referencia SRI2 la fuerza que genera la $fem$ es eléctrica. Además, en la aproximación de que $\gamma {\approx} 1$, las dos generan la misma $fem$**.
```
