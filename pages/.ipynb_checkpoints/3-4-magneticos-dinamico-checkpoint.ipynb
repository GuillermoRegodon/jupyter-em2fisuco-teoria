## Materiales magnéticos en régimen dinámico

Al estudiar los materiales, hemos supuesto que se alcanza un estado de equilibrio de forma suficientemente rápida con respecto a la escala de tiempos de la variación de los campos magnéticos externos. **En este tema, vamos a revisar las ecuaciones de la variación de los dipolos magnéticos** y vamos a resolverlas en algunos casos ilustrativos.

## Ecuación de Landau-Lifshitz

Un dipolo magnético, caracterizado por su momento dipolar magnético $\vec{m}$, inmerso en un campo magnético $\vec{B}$ sufre un torque $\vec{\tau}$ que tiende a orientarlo en la misma dirección de $\vec{B}$.

$$
\vec{\tau} = \vec{m} \times \vec{B}
$$

Por una parte, el torque $\vec{\tau}$ es precisamente la variación del momento angular $\frac{{\rm d}\vec{l}}{{\rm d}t}$, que a su vez está relacionado con el momento dipolar magnético $\vec{m} = -\frac{e}{2m_e} \vec{l} = -\gamma \vec{l}$. Por lo tanto:

$$
\frac{{\rm d} \vec{m}}{{\rm d} t} = - \gamma \vec{m} \times \vec{B}
$$

En un material, no todos los momentos dipolares magnéticos son de la misma magnitud, por lo que cada uno tendrá una dinámica un poco diferente, pero en primera aproximación podemos suponer que todos son iguales. Por lo tanto, para un volumen $\Delta V$ de material,

$$
\frac{{\rm d} \vec{M}}{{\rm d} t} = - \gamma \vec{M} \times \vec{B}
$$

Esta ecuación es conocida como la **ecuación de Landau-Lifshitz**. Vamos a resolverla en algunos casos.

```{note}
**Campo magnético constante $\vec{B} = B_0 \vec{u_z}$**

Estudiando las componentes de la ecuación:

$$
\frac{{\rm d} M_x}{{\rm d} t} &= - \gamma M_y B_0 \\
\frac{{\rm d} M_y}{{\rm d} t} &= \gamma M_x B_0 \\
\frac{{\rm d} M_z}{{\rm d} t} &= 0
$$

La primera conclusión es que $M_z = M_0$ es constante. La segunda se obtiene desacoplando las dos primeras ecuaciones.

$$
\frac{{\rm d^2} M_x}{{\rm d} t^2} &= - \gamma^2 B_0^2 M_x\\
\frac{{\rm d^2} M_y}{{\rm d} t^2} &= - \gamma^2 B_0^2 M_y
$$

Es decir, la imanación en la dirección transversal es una función armónica de frecuencia $\omega_0 = \gamma B_0$, conocida como la **frecuencia de Larmor**. Además, la ecuación original demuestra que el desfase entre las direcciones es $\pi/2$, por lo que la imanación transversal traza una circunferencia en torno al eje $z$.

$$
M_x &= M_1 \cos(\omega_0 t) \\
M_y &= M_1 \sin(\omega_0 t) \\
M_z &= M_0
$$
```

````{note}
**Campo magnético rotatorio $\vec{B} = B_0 \vec{u_z} + B_1 (\cos (\omega t) \vec{u_x} + \sin (\omega t) \vec{u_y})$**

En este caso, las ecuaciones por componentes:

$$
\frac{{\rm d} M_x}{{\rm d} t} &= - \gamma M_y B_0 + \gamma M_z B_1 \sin (\omega t)\\
\frac{{\rm d} M_y}{{\rm d} t} &= \gamma M_x B_0 - \gamma M_z B_1 \cos (\omega t)\\
\frac{{\rm d} M_z}{{\rm d} t} &= \gamma B_1 (M_x \sin (\omega t) - M_y \cos (\omega t))
$$

Aunque a priori, no tenemos un motivo para ello, vamos a realizar la suposición adicional de que $M_z = M_0$ constante.

```{note}
La suposición $M_z = M_0$ se puede justificar por los fenómenos de relajación que presenta el material y que hacen que la imanación acabe atenuándose, de forma que al final solo quedará la imanación inducida por el campo magnético rotatorio. Sin embargo, este efecto no se puede predecir a partir de la ecuación de Landau-Lifshitz, que no tiene en cuenta fenómenos de relajación de la imanación.
```

Con esto, las otras dos ecuaciones se convierten en dos ecuaciones diferenciales de primer orden acopladas, que se pueden desacoplar y resolver:

$$
\vec{M} = M_0 \vec{u_z} + \frac{\gamma M_0}{\omega_0 - \omega}\vec{B_1}
$$

Donde, por simplicidad de la expresión, hemos usado $\vec{B_1} = B_1 (\cos (\omega t) \vec{u_x} + \sin (\omega t) \vec{u_y})$. Hemos obtenido el resultado de que la magnetización sigue al campo magnético en fase, pero solo puede seguirlo si existe el campo magnético longitudinal $B_0 \vec{u_z}$ que genera $M_0$. **Si $M_0 = 0$, el material no puede imanarse con este campo.**

Otro resultado interesante es que la imanación puede entrar en resonancia si la frecuencia de la excitación $\omega$ se acerca a la frecuencia de Larmor $\omega_0$.
````

## Ecuaciones de Bloch

En los materiales imanados, existen fenómenos de relajación que atenúan la imanación. En presencia de una campo magnético constante en la dirección $z$, encontramos dos tipos de relajación:

* **Relajación longitudinal**: Ocurre por la diferencia de energía entre los distintos niveles energéticos. Las vibraciones de la red cristalina o del resto del material tienen una probabilidad de intercambiar energía con estos niveles, con un tiempo característico $\tau_1$. El estado final sería el estado de equilibrio termodinámico característico de los materiales paramagnéticos, con una imanación final $M_0$. Se caracteriza como un sistema de primer orden con un valor final $M_0$:

$$
\frac{{\rm d} M_z}{{\rm d} t} = \frac{M_0 - M_z}{\tau_1}
$$

* **Relajación transversal**: Los materiales que se caracterizan por una imanación $\vec{M} = n \left<\vec{m}\right>$ están constituidos por distintos momentos dipolares magnéticos que en media valen $\left<\vec{m}\right>$. Dado que cada momento dipolar magnético tiene tendencia a precesionar a una frecuencia diferente, el desfase entre estos dipolos magnéticos aumenta. En un tiempo característico $\tau_2$ los dipolos acaban teniendo orientaciones opuestas y la media se anula, $\left<\vec{m}\right> \rightarrow 0$. Se caracteriza como un sistema de primer orden con valor final nulo

$$
\frac{{\rm d} M_{x,y}}{{\rm d} t} = -\frac{M_{x,y}}{\tau_2}
$$

El sistema global, incluyendo el torque, constituye las ecuaciones de Bloch:

$$
\frac{{\rm d} \vec{M}}{{\rm d} t} = - \gamma \vec{M} \times \vec{B} + \left( \begin{array}{c} -\frac{M_x}{\tau_2} \\ -\frac{M_y}{\tau_2} \\ \frac{M_0 - M_z}{\tau_1} \end{array} \right)
$$

````{note}
**Campo magnético constante $\vec{B} = B_0 \vec{u_z}$**

Cuando un material con una determinada imanación se relaja en presencia de una campo magnético constante $\vec{B} = B_0 \vec{u_z}$, el material se puede relajar de varias maneras, dependiendo de los valores de los tiempo de relajación, y como se relacionan con la frecuencia de Larmor, pero la solución más común es que se relaje hacia $\vec{M} \rightarrow M_0 \vec{u_z}$ dando vueltas alrededor del eje $z$.


```{figure} ../images/mat_16.png
:alt: mat_16.png
:width: 400px
Componentes de la imanación en un campo magnético constante según las ecuaciones de Bloch
```
```{figure} ../images/mat_17.png
:alt: mat_17.png
:width: 400px
Evolución de vector imanación en un campo magnético constante según las ecuaciones de Bloch
```

El tiempo de relajación $\tau_1$ siempre es $\tau_1 > \tau_2$, y así se ilustra en las figuras de ejemplo.


````


````{note}
**Campo magnético rotatorio $\vec{B} = B_0 \vec{u_z} + B_1 (\cos (\omega t) \vec{u_x} + \sin (\omega t) \vec{u_y})$**

Cuando la excitación es un campo magnético rotatorio, la solución estacionaria de esta ecuación vectorial es la siguiente:

$$
M_x &= M_1 \cos \varphi \cos (\omega t) + M_1 \sin \varphi \sin (\omega t) = M_1 \cos(\omega t - \varphi) \\
M_y &= M_1 \cos \varphi \sin (\omega t) - M_1 \sin \varphi \cos (\omega t) = M_1 \sin(\omega t - \varphi) \\
M_z &= M_0'
$$

Es decir, la imanación sigue a la excitación con un cierto retraso, y mantiene una imanación transversal rotatoria. Obsérvese que la imanación final longitudinal es menor que en ausencia de campo magnético rotatorio, $M_0'<M_0$, y la constante de tiempo de relajación longitudinal $\tau_1'$ es mayor que en ausencia de campo magnético rotatorio. Además, aparece un retardo entre la fase de la excitación y la magnetización resultante. De todas formas, esta solución es directa, aunque un poco enrevesada de obtener, y de poca utilidad práctica, por lo que no ahondaremos más en ello.

````

## Aplicación a Resonancia Magnética Nuclear

La RMN es una técnica de análisis químico que aprovechar los principios físicos estudiados para descubrir la composición química de una sustancia sin necesidad de descomponer la muestra. Además permite distinguir isómeros, es decir, sustancias químicas que tienen la misma fórmula química pero distintos enlaces. Para ello aprovechamos efectos físicos adicionales o de segundo orden.

La técnica se basa en **aplicar un campo magnético variable con polarización circular que puede ser modelado mediante las ecuaciones de Landau-Lifshitz**. Cuándo alguno de los núcleos entra en resonancia, aparece absorción:

* Una de las aplicaciones más comunes es al **átomo de carbono en sustancias orgánicas**. El núcleo de carbono $^{14}C$ no tiene momento angular neto por tener una número par de protones, pero en el carbono, el $1.07%$ de los núcleos es $^{13}C$ que tiene un espín neto que sí orbita siguiendo esta excitación circular.

* Se observa experimentalmente que **la resonancia del núcleo se ve distorsionada por la presencia de átomos de hidrógeno enlazados con el átomo de carbono**, por lo que, analizando dichas resonancias, podemos saber cuánto átomos de carbono enlazado con 0, 1, 2, 3 o 4 (*en el único caso del metano*) hay en la molécula. **Otros átomos enlazados también modifican la frecuencia de absorción**.

* **Otros núcleos con espín neto pueden ser utilizados**, obteniendo una información muy completa para deducir la estructura química de la molécula.

```{figure} ../images/mat_20.png
:alt: mat_20.png
:width: 400px
Ejemplo de espectro de carbono RMN
```


## Aplicación a Imagen por Resonancia Magnética

La RMI es la técnica médica que ha revolucionado la diagnosis médica. Antes de su diseño y utilización, teníamos las radiografías para analizar el tejido sólido (*Huesos principalemente, además de otros objetos extraños como pueden ser piedras renales o objetos ingeridos*). Sin embargo, **para analizar tejido blando, era necesaria una operación quirúrjica y observación directa. La RMI se utiliza para estudiar principalmente el átomo de hidrógeno, presente en todos los tejidos blandos del cuerpo.**


```{figure} ../images/mat_21.png
:alt: mat_21.png
:width: 400px
Rayos X vs IRM
```


La técnica se basa en **aplicar un campo magnético variable con polarización circular que puede ser modelado mediante las ecuaciones de Bloch**. La técnica es extremadamente compleja e incluye hitos científicos e ingenieriles en muchas de las facetas que la componen: En particular, se necesita un electroiman de campo magnético suficientemente intenso (*$B \approx 3T$*) con una región aproximadamente constante con un pequeño gradiente perfectamente controlado lo suficientemente grande para que quepa una persona. El electroiman necesita una corriente eléctrica tan elevada que está generado por un metal en estado de superconductor para que las perdidas por resistencia del material sean nulas. ¡La mayor parte de la energía que consume el aparato de hecho se invierte en mantener el electroiman suficientemente frío!

El proceso de captura de la imagen tiene los siguientes pasos:

* Inicialmente, cuando el tejido se introduce en el campo magnético (*La persona es colocada en la camilla del aparato*) **los núcleos de hidrógeno** (*y todos los demás, aunque no participan en la medida*) **se alinean con el campo magnético**.

* **Se emite un pulso de radiación RF con polarización circular**. Dado que el tejido está en un campo magnético con un gradiente, cada plano del cuerpo está sometido a un campo magnético ligeramente diferente, por lo que **sólo el tejido en un plano está sometido a resonancia** y absorve eficientemente la radiación (*De hecho, se calienta ligeramente. Para no producir fiebre en la persona, el número de medidas seguidas es limitado.*).

* La energía absorvida por el tejido de dicho plano se acumula en energía potencial magnética que saca a los núcleos de hidrógeno de dicho alineamiento con el campo magnético. **Se elige aplicar RF hasta que los núcleos de hidrógeno se alinean perpendiculares u opuestos al campo magnético**, en función de la medida que se quiera realizar.

* Una vez desaparece el pulso de RF, **los núcleos de hidrógeno empiezan a orbitar según las ecuaciones de Bloch**. Dicha órbita se puede medir utilizando bobinas longitudinales o transversales receptoras. **Para poder diferenciar distintas líneas, se añade un pequeño gradiente transversal constante**, de tal forma que cada linea orbita a un ritmo distinto. La señal que se registra es única, pero con herramientas matemáticas como la transformada de Fourier se pueden diferenciar después.

* **La corriente inducida medida en las bobinas presenta dos decaimientos, el longitudinal o *T1* y el transversal o *T2*. Lo que se registra el la constante de tiempo del decaimiento que se obtiene.**

```{figure} ../images/mat_22.png
:alt: mat_22.png
:width: 600px
Gradiente en z y gradiente transversal
```

El decaimiento que sufren los espines desalineados en su proceso de vuelta al estado de alineamiento, tal y como predicen las ecuaciones de Bloch, está relacionado con diversos efectos de segundo orden.

1. **Los dipolos magnéticos se alinean con el campo magnético externo por las colisiones con las moléculas del entorno**, por lo que se llama **relajación espín-red**. La energía transferida a la red es la que calienta ligeramente el tejido y se produce un decaimiento +T1* a un ritmo de decaimiento $\tau_1$.
2. Los dipolos magnéticos obitan a la misma frecuencia para un determinado campo magnético. Sin embargo, **las irregularidades del campo magnético externo junto con la influencia de los dipolos vecinos** que pueden incluso ser otros átomos con espín neto distintos al hidrógeno, hacen que **los espines se desincronicen entre ellos**, produciendo un decaimiento *T2* llamado r**elajación espín-espín** a un ritmo distinto $\tau_2$.

Dado que ambos tiempos de decaimiento dependen del entorno del espín, cada sustancia tiene unos tiempos *T1* y *T2* característicos. Por ejemplo, el tejido graso aparece brillante en *T1* pero oscuro en *T2*, mientras que el líquido cefaloraquídeo (*esencialmente agua*) brilla en *T2*.
